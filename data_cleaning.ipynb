## correlation analysis

```python
correlation_features = [
    "Age",
    "Tenure",
    "Usage Frequency",
    "Support Calls",
    "Payment Delay",
    "Total Spend",
    "Last Interaction",
    "Churn"
]

correlation_matrix = df[correlation_features].corr()

display(correlation_matrix)
```

## churn correlation ranking

```python
churn_corr = (
    df[correlation_features]
    .corr()["Churn"]
    .drop("Churn")
    .sort_values(ascending=False)
)

print("Correlation with Churn:")
display(churn_corr.to_frame("Correlation"))
```

## create age groups

```python
df["Age Group"] = pd.cut(
    df["Age"],
    bins=[0, 25, 35, 45, 55, 100],
    labels=[
        "18-25",
        "26-35",
        "36-45",
        "46-55",
        "56+"
    ]
)

display(
    pd.crosstab(
        df["Age Group"],
        df["Churn"],
        normalize="index"
    ).mul(100).round(2)
)
```

## create tenure groups

```python
import numpy as np
df["Tenure Group"] = pd.cut(
    df["Tenure"],
    bins=[0, 12, 24, 36, 48, 60, np.inf],
    labels=[
        "0-12 months",
        "13-24 months",
        "25-36 months",
        "37-48 months",
        "49-60 months",
        "60+ months"
    ]
)

tenure_churn = pd.crosstab(
    df["Tenure Group"],
    df["Churn"],
    normalize="index"
) * 100

display(tenure_churn)
```

## create customer risk features

```python
df["Payment_Risk"] = pd.cut(
    df["Payment Delay"],
    bins=[-np.inf, 7, 15, 30, np.inf],
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High"
    ]
)
```

```python
df["Support_Risk"] = pd.cut(
    df["Support Calls"],
    bins=[-np.inf, 3, 6, 9, np.inf],
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High"
    ]
)
```

```python
df["Engagement_Risk"] = pd.cut(
    df["Usage Frequency"],
    bins=[-np.inf, 5, 15, 25, np.inf],
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High"
    ]
)
```

```python
display(
    df[
        [
            "Payment_Risk",
            "Support_Risk",
            "Engagement_Risk",
            "Churn_label"
        ]
    ].head()
)
```

## analyze payment risk vs churn

```python
payment_risk_churn = pd.crosstab(
    df["Payment_Risk"],
    df["Churn"],
    normalize="index"
) * 100

display(payment_risk_churn)
```

## business summary table

```python
business_summary = df.groupby("Churn").agg({

    "Age": "mean",
    "Tenure": "mean",
    "Usage Frequency": "mean",
    "Support Calls": "mean",
    "Payment Delay": "mean",
    "Total Spend": "mean",
    "Last Interaction": "mean"

}).round(2)

business_summary.index = [
    "Retained",
    "Churned"
]

display(business_summary)
```
